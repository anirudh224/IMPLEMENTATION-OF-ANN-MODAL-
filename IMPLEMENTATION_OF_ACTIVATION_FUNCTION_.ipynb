```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras import  layers
```

```python
import numpy as np
import pandas as pd

# Set random seed for reproducibility
np.random.seed(42)

# Generate 15 rows of dummy data for the specified characteristics
soil_moisture = np.round(np.random.uniform(10.0, 80.0, 15), 2)  # Percentage
temperature_c = np.round(np.random.uniform(15.0, 38.0, 15), 1)  # Celcius
sunlight_hours = np.round(np.random.uniform(2.0, 12.0, 15), 1)  # Hours

# Logic for needs_water: 1 if soil moisture is low (< 35) or temperature is high (> 30), else 0
needs_water = ((soil_moisture < 35) | (temperature_c > 30)).astype(int)

data = {
    'soil_moisture': soil_moisture,
    'temperature_c': temperature_c,
    'sunlight_hours': sunlight_hours,
    'needs_water': needs_water
}

df = pd.DataFrame(data)
display(df)
```

```python
df.columns
```

```python
X=df[['soil_moisture','temperature_c','sunlight_hours']]
y=df[['needs_water']]
```

```python
X_min = X.min()
X_max = X.max()
X_scaled = (X - X_min) / (X_max - X_min)
```

```python
X_scaled
```

```python
X_train,X_test,y_train,y_test=train_test_split(X_scaled,y,test_size=0.25,random_state=42,stratify=y)
```

```python
model = keras.Sequential([
    layers.Input(shape=(X_train.shape[1],)),
    layers.Dense(8, activation='relu'),
    layers.Dense(3, activation='softmax')
])
```

firstly us sigmoid when output is single layer if there is multiple then use soft max


```python
model.compile(optimizer='SGD', loss='categorical_crossentropy', metrics=['accuracy'])
```

if we are difinteating in many terms than use categorical cross entropy

```python
history = model.fit (X_train.values , y_train.values,
validation_data =(X_test.values,y_test.values),epochs=100, batch_size=4, verbose=1)
```
